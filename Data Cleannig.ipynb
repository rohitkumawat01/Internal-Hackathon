```python
df = df.drop_duplicates()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "")
    .str.replace(")", "")
    .str.replace("/", "_")
)
```

```python
df["order_placed_at"].head()
```

```python
df["order_placed_at"] = pd.to_datetime(
    df["order_placed_at"],
    format="%I:%M %p, %B %d %Y",
    errors="coerce"
)
```

```python
df["order_placed_at"].dtype
```

```python
df["order_date"] = df["order_placed_at"].dt.date
df["order_year"] = df["order_placed_at"].dt.year
df["order_month"] = df["order_placed_at"].dt.month
df["order_day"] = df["order_placed_at"].dt.day
df["order_hour"] = df["order_placed_at"].dt.hour
```

```python
df["distance"].head(20)
```

```python
df["distance_km"] = (
    df["distance"]
    .str.extract(r"([\d.]+)")[0]
    .astype(float)
)
```

```python
df["item_count"] = (
    df["items_in_order"]
    .str.extract(r"(\d+)")[0]
    .astype(float)
)
```
